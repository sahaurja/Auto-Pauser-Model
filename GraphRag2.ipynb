```python
from fastcoref import FCoref
import torch
device = "cuda" if torch.cuda.is_available() else "cpu"
model = FCoref(device = device)
```

For each cluster, combine into one canonical representative

```python
import re
```

```python
from collections import Counter
#optimized version
def get_cannonical_rep(doc_ents, my_clus):
  #store the type of each cluster
  type_counts = Counter()
  #representative for the cluster
  max_word = ""
  #whether the max_word is an entity
  max_is_ent = False
  #all non-representative words in that cluster
  aliases = set()
  #all words that cannot be representative or alias
  not_allow = set(["the", "it", "its", "we", "that", "they", "this", "these", "those", "he", "she"]) #set for O(1) search
  for og_word in my_clus:
    #remove any 's in the original word
    cleaned = re.sub(r"’s","",og_word)
    if(cleaned == ""): #empty string cannot be representative or alias
      continue
    #check if the original word is an entity
    og_is_ent = og_word in doc_ents
    if(og_is_ent):
      type_value = doc_ents[og_word]
      type_counts[type_value] += 1
    #check if it is one of the unallowed common nouns
    if(cleaned.lower() in not_allow):
      aliases.add(cleaned)
      continue
    #check whether or not max_word should be replaced
    both_are_ent = max_is_ent and og_is_ent
    only_new_ent = (not max_is_ent) and og_is_ent
    neither_ent = (not max_is_ent) and (not og_is_ent)
    do_replace = (both_are_ent and len(cleaned) > len(max_word)) or (only_new_ent) or (neither_ent and len(cleaned) > len(max_word))
    #act on value of do_replace
    if(do_replace):
      if(both_are_ent):
        aliases.add(max_word)
        max_word = cleaned
      elif(only_new_ent):
        if(max_word != ""):
          aliases.add(max_word)
        max_word = cleaned
        max_is_ent = True
      else:
        if(max_word != ""):
          aliases.add(max_word)
        max_word = cleaned
    else:
      if(cleaned != max_word): #max_word is not an alias of itself
        aliases.add(cleaned)
  #return the final result
  ent_type = type_counts.most_common(1)[0][0] if type_counts else None
  return {"representative": max_word, "aliases": aliases, "type": ent_type}
```

**NER+Coref+Indexing for sections_reg**

```python
#replace any non-ascii characters
def normalize_text(text):
    text = text.replace('\u00a0', ' ') # non-breaking space -> regular space
    text = text.replace('\u2019', "'") # curly apostrophe -> straight apostrophe
    text = text.replace('\u2018', "'")
    text = text.replace('\u201c', '"').replace('\u201d', '"') # curly quotes -> straight
    text = text.replace('\n', '') # removes all newline characters
    text = re.sub(r'\s+', ' ', text)
    return text
```

```python
sections_reg = ["Item 1", "Item 1A", "Item 1B", "Item 1C", "Item 2", "Item 3", "Item 4", "Item 9", "Item 9A", "Item 9B", "Item 9C", "Item 10", "Item 11", "Item 12", "Item 13", "Item 14", "Item 16"]
all_rep = []
all_mention_index = {}

for curr_section in sections_reg:
  text = normalize_text(tenk[curr_section])
  doc = nlp(text)
  doc_ents = {ent.text: ent.label_ for ent in doc.ents} #dict of each entity with its type
  #get clusters
  preds = model.predict(texts = text)
  clusters_strings = preds.get_clusters()
  cluster_spans = preds.get_clusters(as_strings = False) #get start and end indices for each entity (named or unnamed)

  section_reps = []
  mention_index = []

  #local_cluster_id is local to specific representative group
  for local_cluster_id, (cluster_string, cluster_span) in enumerate(zip(clusters_strings, cluster_spans)):
    #deal with text values of entities
    rep = get_cannonical_rep(doc_ents, cluster_string) #get all organized clusters for that item
    section_reps.append(rep) #consolidate all the organized clusters
    #deal with postion values of entities
    for (start_pos, end_pos) in cluster_span:
      mention_index.append((start_pos, end_pos, local_cluster_id)) #local_cluster_id is associated with the position of the {rep} in all_rep

  all_rep.append(section_reps)
  all_mention_index[curr_section] = mention_index
```

```python
#remove title headers to clean entities like Human CapitalThe Company
#if in a representative, there is a word with a lowercase letter followed by a capital letter, only keep the part from the second capital letter in that word, inclusive...
#so instead of Human CapitalThe Company, keep The Company


def clean_representative(text):
    # Matches a lowercase letter followed by an uppercase letter
    # and captures everything from that uppercase letter to the end
    match = re.search(r'[a-z]([A-Z].*)', text)

    if match:
        return match.group(1) # Returns the captured part
    return text # Returns original text if no match is found

# Example usage:
for item in all_rep:
  for cluster in item:
    cluster["representative"] = clean_representative(cluster["representative"])
```

**Handle Normalizing and Manual References**

```python
FILER_NAME = "Apple Inc." #official company name
FILER_ALIASES = {"the company", "the company's" "the registrant", "we", "us","our", "apple", "company"}

def normalize_key(text):
  key = text.lower().strip()
  key = re.sub(r"[.,]", "", key) #remove , and .
  key = re.sub(r'\b(inc|incorporated|corp|corporation|co|ltd|llc)\b', '', key) #remove extra company title info
  key = re.sub(r'\s+', ' ', key).strip() #make singular
  return key

FILER_KEY = normalize_key(FILER_NAME) #apple
```

**Global Merge Across Different Items**

```python
global_entities = {}
for item_name, reps in zip(sections_reg, all_rep):
  for rep in reps:
    normalized_key = normalize_key(rep["representative"]) #normalize for checking purposes
    key = ""
    #get the global key value
    if(normalized_key in FILER_ALIASES or normalized_key == FILER_KEY):
      key = FILER_KEY #key = "apple"
    else:
      key = normalized_key #key is the normalized version of itself
    #add to global entities if not already there
    if (key not in global_entities):
      global_entities[key] = {
          "representative" : FILER_NAME if key == FILER_KEY else rep["representative"], #original version, not normalized
          "aliases" : set(rep["aliases"]),
          "type_votes": Counter(rep["type"]) if rep["type"] else Counter(),
          "sections": {item_name}
      }
    #else update global entities
    else:
      entry = global_entities[key]
      entry["aliases"].update(rep["aliases"])
      if(rep["type"]):
        entry["type_votes"][rep["type"]] += 1
      entry["sections"].add(item_name)

for entry in global_entities.values():
  entry["type"] = entry["type_votes"].most_common(1)[0][0] if entry["type_votes"] else None
  del entry["type_votes"]

for global_ind, (key, entry) in enumerate(global_entities.items()):
  entry["global_id"] = f"ent_{global_ind}"
```

```python
key_to_global_id = {key: entry["global_id"] for key, entry in global_entities.items()} #map normalized key to global id
global_id_to_name = {entry["global_id"] : entry["representative"] for entry in global_entities.values()} #map global id to key's original value
```

**Convert local_cluster_id to global_cluster_id**

```python
local_global_cluster = {} #item name -> {local id : global id}

for item_name, reps in zip(sections_reg, all_rep):
  mapping = {}
  for local_id, rep in enumerate(reps):
    normalized_key = normalize_key(rep["representative"]) #normalize representative value
    key = ""
    #get the global key value
    if(normalized_key in FILER_ALIASES or normalized_key == FILER_KEY):
      key = FILER_KEY #key = "apple"
    else:
      key = normalized_key #key is the normalized version of itself
    mapping[local_id] = key_to_global_id[key]
  local_global_cluster[item_name] = mapping
```

**Do the Substitution**

```python
def apply_sub (text, replacements):
  #do in reverse order, as to not mess up indexing
  for start, end, rep in sorted(replacements, key = lambda r : r[0], reverse = True):
    text = text[:start] + rep + text[end:]
  return text
```

```python
resolved_items = {}

for item_name in sections_reg:
  text = normalize_text(tenk[item_name])
  mention_index = all_mention_index[item_name]
  cluster_local_global = local_global_cluster[item_name]
  #store the replacements
  replacements = []
  for start_pos, end_pos, local_cluster_id in mention_index:
    global_id = cluster_local_global[local_cluster_id]
    replacements.append((start_pos, end_pos, global_id_to_name[global_id]))
  resolved_items[item_name] = apply_sub(text, replacements)
```
