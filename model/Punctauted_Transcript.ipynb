```python
import spacy
```

```python
nlp = spacy.load("en_core_web_md") #english model
```

```python
#iterate along and check for all vectors
all_pauses = [] #too different
all_cont = [] #similar enough
#threshhopld kind of randomly chosen for now
thresh = 0.7
for i in range(len(sentence_list)-1):
  doc1 = nlp(sentence_list[i])
  doc2 = nlp(sentence_list[i+1])
  sim = doc1.similarity(doc2)
  if(sim < thresh):
    all_pauses.append({"snippet1": doc1, "snippet2": doc2, "sentence 1": i, "sentence 2": i+1, "similarity": sim})
  else:
    all_cont.append({"snippet1": doc1, "snippet2": doc2, "sentence 1": i, "sentence 2": i+1, "similarity": sim})
```
